## 1. Introduction

### Background:
Predicting the results of soccer matches has long been a challenging task due to the dynamic nature of sports and the numerous factors influencing results. From in-game events like goals and red cards to pre-game metrics such as team form and historical performance, soccer analytics presents a rich and complex domain for machine learning applications. Even though how rich are the inputs, these features are never enough for a robust model. The outcomes are affected by so many aspects which are not parameterized by the inputs. Due to this nature of the sports competitions it has always been a challenge to model these problems.

This project addresses the intersection of sports analytics and live betting by developing a data-driven framework for predicting match outcomes and identifying optimal betting opportunities. By leveraging real-time data and machine learning models, the study aims to replicate market behavior and outperform conventional betting strategies.

### Objectives
The primary objectives of this project are:
Outcome Prediction: Accurately predict match results (home win, draw, or away win) using advanced machine learning models.
Betting Strategy Optimization: Identify the most profitable moments to place bets during a match, considering real-time odds and market probabilities.

### Methods Overview
Data Preparation:
- Sorting and grouping match data by time order to preserve chronological integrity.
- Forward-filling and imputing missing values to create a consistent dataset.

Feature Engineering:
- Deriving team-specific metrics such as form, streaks, and weighted performance.
- Calculating advanced statistics like shot efficiency, expected goals, and game pressure for each team.

Modeling Approaches:
- Classification models (Random Forest, Gradient) for predicting match outcomes.
- Regression models for predicting goal differences as a basis for betting decisions.
- Time-series cross-validation is applied to tune both classification and regression models.

Evaluation Metrics:
- Ranked Probability Score (RPS): Measures the quality of probabilistic predictions.
- Expected Return: Assess the total profit from simulated bets and identify profitable patterns.

## 2. Literature Review
Predicting soccer match outcomes and developing betting strategies have long been of significant interest within the fields of sports analytics, machine learning, and predictive modeling. Recent research has explored various methods to improve prediction accuracy, focusing on statistical models, machine learning approaches, and domain-specific feature engineering.

One of the earliest and most influential approaches to predicting soccer outcomes was based on statistical modeling. Maher (1982) introduced a Poisson regression model to predict match results by estimating the number of goals scored by each team. This approach relied on historical match data, assuming that goals follow a Poisson distribution. The model demonstrated reasonable predictive power, especially in terms of predicting the number of goals rather than binary outcomes like win, loss, or draw.

Berrar et al. (2019) highlighted the importance of feature engineering in machine learning models for soccer prediction. Their study introduced two novel feature engineering methods: “recency feature extraction” and “rating feature learning.” The recency feature extraction method captures a team’s recent form by considering performance metrics over a sliding time window. In contrast, rating feature learning assigns dynamic ratings to teams based on their attacking and defensive capabilities, which are updated after each match.

Horvat and Job (2020) conducted an extensive review of machine learning algorithms applied to sports outcome prediction, analyzing over 100 papers. They observed that neural networks, particularly deep learning models, were among the most frequently used algorithms. However, they noted that traditional models like logistic regression and SVM still performed competitively when combined with effective feature selection and extraction.

Baboota and Kaur (2019) explored the use of random forests and gradient boosting for predicting outcomes in the English Premier League. Their work emphasized the role of feature selection in improving model performance. By selecting key features such as goal differences, team strength ratings, and recent match statistics, their models outperformed several baseline approaches. However, their models, while accurate, were unable to consistently outperform bookmakers’ odds due to the inherent unpredictability of soccer matches. The weighted streak and team form features from this study are integrated into our model and discussed in the methodology section.

Horvat et al. (2020) discussed the importance of modeling dynamic odds in live betting. They noted that in-game events such as goals, red cards, and substitutions significantly impact the odds offered by bookmakers. Traditional machine learning models, which rely on static pre-match data, are insufficient for live betting scenarios. Instead, models need to incorporate streaming data and update predictions continuously throughout the match.

## 3. Data Overview

### Dataset Description
The dataset contains detailed information on soccer matches, including real-time match statistics, betting odds, and metadata. It is designed to support dynamic, time-sensitive analysis for predicting match outcomes and optimizing live betting strategies.

Key features of the dataset include:
- Scope: Covers matches before November 1, 2024, for training and those after this date for testing.

Columns:
- Metadata: Includes match identifiers (fixture_id), start times (match_start_datetime), and specific time points (minute, second).
- Odds Data: Raw betting odds (1, X, 2) and their normalized probabilities.
- Match Statistics: Offensive metrics (e.g., shots on target), defensive metrics (e.g., tackles), and event-based measures (e.g., goals, corners, red cards).
- The dataset contains about 50,000 rows, with dynamically updated features capturing both offensive and defensive aspects of the game.


### Preprocessing
To ensure data quality and consistency, the following preprocessing steps were applied:

Sorting and Forward-Filling:
- Data is sorted by match time to preserve chronological order.
- Missing values within matches are forward-filled to maintain continuity, while initial missing values were replaced with zeros.

Odds Normalization:
- Raw odds are converted into probabilities and normalized to account for bookmaker adjustments, ensuring these probabilities summed to one.

Filtering:
- Suspended or stopped matches are excluded, as the associated odds and statistics are deemed unreliable.

Feature Integration:
- Derived features like goal difference, attack and defense scores, and team form metrics are added to enrich the dataset.
- Team-specific historical data is incorporated to capture performance trends.
- These steps resulted in a comprehensive dataset of around 150 features per row, ready for predictive modeling and betting simulations.


### Feature Engineering
This is a critical step for this project since it directly affects the predictive capabilities of the machine learning models. The logic behind the features extracted and selected directly from the raw dataset relies on the content - knowledge or literature review. There are 2 main categories under the feature engineering phase.

#### Processed Features
The following features are highly engineered using domain knowledge and mathematical transformations. These features aim to capture the dynamic, tactical, and psychological elements of soccer matches. 

All the features below are shown for the home teams. However, all of them can be calculated for away teams too.

- Goal Difference: Serves as a feature assessing team’s dominance. = (Goals - home) - (Goals - away)

- Shot Efficiency: Measures how efficiently a team converts total shots into goals. Consider that higher efficiency reflects better finishing quality and is a strong indicator of offensive performance. = (Goals - home) / Shots Total - home

- Cross Success Rate: The proportion of effective crosses over the total attempts. Indicates the ability to create scoring opportunities through effective wing play. = (Accurate Crosses - home) /  Total Crosses - home

- Shot Distribution (Close Shots): The proportion of the shots taken inside the box. Close range shots have more goal expectations. = (Shots Insidebox - home) / Shots Total - home

- Attack to defense Ratio: Compares offensive and defensive effectiveness. Teams with a higher ratio are more likely to dominate the game. = (Dangerous Attacks + Goal Attempts) / (Tackles + Interceptions + Shots Blocked)

- Expected Goals: A weighted measure of scoring opportunities. = (0.3 * Shots on Target - home) + (0.15 * Shots Insidebox - home) + (0.05 * Shots Outsidebox - home)

- Attack Momentum: Measures the offensive pressure over the last 10 minutes. Indicates the closeness to the score. = Dangerous Attacks + (Goal Attempts * 2) + (Successful Dribbles * 0.5)

- Defensive Pressure: Quantifies defensive actions relative to the opponent’s attacks. = ((Tackles * 0.2) + (Interceptions * 0.15) + (Shots Blocked * 0.1)) / Opponent’s Dangerous Attacks

- Set Piece Thread: Weighted contribution of set pieces to goal-scoring opportunities. = (Corners * 0.15) + (Free Kicks * 0.3) + (Penalties * 0.85)

- Psychological Momentum: Measures the current psychological state based on recent events which affects the players mood. = Goals - Red Cards - Injuries / 2

- Weighted Streak: A weighted average of recent match outcomes, more recent matches have higher weights.

### Directly Selected Features
These features are used to construct the machine learning models directly from data or simple processes.

## 4. Methodology
In this section, methodology employed to develop the betting strategy is discussed. The approach is designed with the key considerations of making a single and accurate bet per match at an optimal time while using appropriate data and avoiding forward-seeing. To use the available data for training the model, some preprocessing steps are taken. Unusable data points which have matches suspended or stopped are dropped. Missing values within individual matches are filled using forward filling and the remaining initially null values are filled as 0. Probabilities for the outcomes are calculated from the given odds in the dataset. Available length of each match is checked and matches with less than 40 minutes available are dropped for improving the accuracy of the model. Then the team features are created, named as interested_col for previous matches for teams having at least 2 matches. After inspecting the literature, we decided to integrate weighted streak and team form features from the reviewed literature to our model. Ratios for the features that are already present in the dataset are calculated and the home team ratios are added. Then the metrics after a 10-minute mark are selected for use since these cumulative metrics will not be meaningful for the first minutes of the game while they are initialized. Null values for the features are checked for the usability of these features. For features having more null values, 3 prior games are used instead of 2 to reduce features with high null values. The given data is split up into test and training data by assigning the matches starting from November 1, 2024, to test, and remaining to training and tuning. Ranked probability scores are used to tune the classification algorithms.

For the first betting method, which is for classification models, the first 10 minutes are considered as initialization of matches and their features, so they are excluded from the betting window. The model then predicts the outcomes of these matches and calculates the expected profit using odds and betting probabilities. A bet with the highest expected profit is selected, if it is positive and also is the most expected outcome. If no bet fulfills this condition, then no bet is placed. Two versions of classification features are tested. In version 1, odds and probabilities from matchmaker are dropped and in version 2, they are used. Version 1 tests without odd information resulted in less overfitting to market odds and higher ability to exploit them, resulting in higher profits.

For the second betting method, which is for regression models, the first 10 minutes are considered as initialization and they are excluded from the betting window similar to first betting method. In the second method, regressors are used to predict goal differences. The betting strategy is then constructed as such: betting on home win if the predicted value is greater than 1, betting on away win if the predicted value is less than -1 and betting on a draw if the predicted value is very close to zero. Bets are made only if a clear prediction is made, and profit or loss is calculated based on the match result.

#### Discussion

Even though accuracy metrics are a little better in RF with odd information compared to the one without odd information, profits are much lower. This indicates that using odds as features to predict match outcomes causes overfitting to odds and not being able to exploit their inefficiencies.

The classification based approach consists of 3 models and the random baseline predictions model. The Random Forest Classification Without Odds data results in the most effective strategy for profitable betting. The profit is 26.8 units and the return is 33.94%. High profit and return ratio indicates that the model captures the underlying market inefficiencies at most. The absence of odds information saves the model from overfitting to bookmaker probabilities. Using Random Forest with classification results in better solutions for every metric compared to Gradient Boosted Decision Trees as expected. 

On the other hand, the inclusion of odds information (RF with Odds Data) results in a slight improvement in Ranked Probability Score with the value (0.1549) at the cost of profitability 7.82 units, suggesting that market odds align closely with expected outcomes. While this model is still effective but the generalization ability is less than the previous model and the fact that expected probability is reduced drastically, this model is not the most beneficial option.

Gradient Based Decision Trees Classification model, despite achieving an okey RPS (0.2179), suffered significant losses (-10.2 units). This result underscores the model’s sensitivity to feature noise and its inability to balance accuracy and profitability effectively.

The random prediction -baseline- results, with the highest RPS score 0.2626 and the highest loss -29.9 units, highlights that the random prediction is inefficient in this scenario. This specific problem needs special features tailored with the domain knowledge to achieve a sensible result.

## 7. Conclusion
The study focuses on constructing machine learning models to predict the results of the soccer matches and designing a live betting strategy. Detailed match statistics data, engineered features, and dynamic odds are used to build models such as Random Forest Regressor/Classifier and Gradient Boosted Decision Trees Regressor/Classifier. Models evaluated for their predictive accuracy, ranked probability scores, and their expected profitability in simulated betting scenarios. 

Among the experiments with the constructed models, Random Forest classifier without using odds data achieved the best profit of 26.81 units and a 33.94% return to bet ratio. This model leveraged engineered features like psychological momentum, dominance index, and attack-to-defense ratios to exploit inefficiencies in betting markets, outperforming all other models. Including odds in the feature set improved prediction accuracy (RPS) but reduced profitability, as seen with Random Forest (With Odds), which highlighted the trade-off between alignment with market expectations and the ability to identify high-value bets.

Between 2 regression models, the Gradient Boosted Decision Trees regressor demonstrated the highest profitability 10.44 units and 12% return to bet ratio, shows clearly that the regression models lag behind classification models in the overall profitability. 

A consistent challenge across all models was predicting draws (X), which exhibited lower precision and recall compared to home or away wins. This indicates potential gaps in feature representation for such outcomes. Additionally, the poor performance of random baselines underscores the necessity of structured modeling and effective feature engineering.

In conclusion, this study explains the potential of machine learning algorithms to inform profitable betting strategies and provide actionable insights into soccer analytics. The outcomes of the experiments are evidence that the adopted techniques can be generalized to utilize for other sport competitions. By carefully balancing prediction accuracy and market inefficiencies, data-driven approaches like this can serve as a valuable tool for decision-making in sports analytics and betting markets.
